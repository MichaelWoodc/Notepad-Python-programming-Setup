```python
from IPython.display import HTML

HTML("""
<div style="font-family: monospace; max-width: 700px;">

  <label>Input:</label><br>
  <input id="argv_input" type="text" style="width: 600px;"
         placeholder="paste your example input here">

  <br><br>

  <button id="argv_generate"
          style="background: #2196F3; color: white; padding: 6px 12px; border: none;">
      Generate sys.argv
  </button>

  <button id="argv_copy"
          style="background: #4CAF50; color: white; padding: 6px 12px; border: none; margin-left: 10px;">
      Copy
  </button>

  <br><br>

  <label>sys.argv:</label><br>
  <textarea id="argv_output" style="width: 600px; height: 20px;"></textarea>

</div>

<script>

document.getElementById("argv_generate").onclick = function() {
    let raw = document.getElementById("argv_input").value.trim();
    let parts = raw === "" ? [] : raw.split(/\\s+/);
    let argv = ["", ...parts];
    document.getElementById("argv_output").value = JSON.stringify(argv);
};

document.getElementById("argv_copy").onclick = function() {
    let text = document.getElementById("argv_output").value;
    navigator.clipboard.writeText(text);
};

</script>
""")
```

### This is an example of how to write your code to run it in vscode, OR notepadd++ or any other way with command line arguments, i.e. powershell or CMD

```python
import sys

# If no arguments were passed (only the script name is present)
if len(sys.argv) == 1:
    # Hard‑code fallback values
    sys.argv = ["", "1", "2", "3", "4", "5"]

# Now you can safely use sys.argv[1:]
print(sys.argv)
```
